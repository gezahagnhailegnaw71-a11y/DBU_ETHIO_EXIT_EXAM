# 2. Why Numerical Analysis?

Many mathematical problems have exact formulas. But in practice, we often face problems where:

- an exact formula does not exist;
- an exact formula exists but is too expensive to use;
- the data are approximate;
- the calculation is too large to perform by hand;
- the answer must be computed repeatedly.

For example:

$$
x^2=2
$$

has the exact positive solution

$$
x=\sqrt{2}.
$$

But a computer stores only a finite approximation such as

$$
1.41421356237\ldots
$$

So the practical question becomes:

> **How do we compute an approximation, how accurate is it, and when should we stop?**

That is the spirit of Numerical Analysis.

# 3. The Numerical Analysis Workflow

A useful way to think about almost every numerical method is:

$$
\boxed{
\text{Problem}
\rightarrow
\text{Mathematical Model}
\rightarrow
\text{Numerical Method}
\rightarrow
\text{Computer Algorithm}
\rightarrow
\text{Check the Answer}
}
$$

### Example

Suppose we want to solve

$$
x^2-2=0.
$$

- **Problem:** Find a number whose square is 2.
- **Model:** Solve $$f(x)=x^2-2=0.$$
- **Method:** Use an iterative root-finding algorithm.
- **Computer algorithm:** Repeat an update formula.
- **Verification:** Check the error or the residual.

This same workflow will appear throughout Chapters 1–3.

# 14. Exact Value vs Numerical Approximation

The exact positive solution of

$$
x^2=2
$$

is

$$
x=\sqrt{2}.
$$

Suppose we use

$$
x_{\text{approx}}=1.4.
$$

The approximation is useful only if we understand its error.

### Absolute error

$$
E_{\text{abs}}
=
|x_{\text{exact}}-x_{\text{approx}}|.
$$

### Relative error

$$
E_{\text{rel}}
=
\frac{|x_{\text{exact}}-x_{\text{approx}}|}
{|x_{\text{exact}}|}.
$$

````python
x_exact = np.sqrt(2)
x_approx = 1.4

absolute_error = abs(x_exact - x_approx)
relative_error = absolute_error / abs(x_exact)

print(f"Exact value    = {x_exact:.10f}")
print(f"Approximation  = {x_approx:.10f}")
print(f"Absolute error = {absolute_error:.10f}")
print(f"Relative error = {relative_error:.10f}")
print(f"Percent error  = {100*relative_error:.6f}%")
````

# 15. Iteration: Improve an Approximation Repeatedly

A central idea in Numerical Analysis is:

> Start with an approximation, improve it, and repeat.

For

$$
\sqrt{2},
$$

consider

$$
x_{k+1}
=
\frac{1}{2}
\left(
x_k+\frac{2}{x_k}
\right).
$$

Start with

$$
x_0=1.
$$

````python
x = 1.0

print(f"{'k':>3} {'x_k':>15} {'absolute error':>18}")
print("-" * 40)

for k in range(6):
    error = abs(np.sqrt(2) - x)
    print(f"{k:3d} {x:15.10f} {error:18.6e}")
    x = 0.5 * (x + 2/x)
````

### Numerical intuition

Observe three things:

1. The approximations move toward a stable number.
2. The error becomes smaller.
3. The same update formula is used repeatedly.

This is the basic pattern behind:

- Fixed-Point iteration;
- Newton's method;
- Jacobi iteration;
- Gauss–Seidel iteration.

# 16. When Should an Iteration Stop?

We cannot iterate forever.

A practical algorithm needs a **stopping rule**.

Common choices include:

$$
|x_{k+1}-x_k|<\text{tolerance},
$$

or

$$
|f(x_k)|<\text{tolerance}.
$$

A good program should also have a maximum number of iterations to prevent endless looping.

````python
x = 1.0
tol = 1e-8
max_iter = 50

for k in range(max_iter):
    x_new = 0.5 * (x + 2/x)

    if abs(x_new - x) < tol:
        x = x_new
        print("Converged.")
        print("Iterations =", k + 1)
        break

    x = x_new

print(f"Approximation = {x:.12f}")
print(f"Residual      = {abs(x**2 - 2):.3e}")
````

### Important idea

A stopping rule is part of the numerical method.

The computer does not “know” when the answer is good enough unless we define a criterion.

# 17. Error and Residual Are Different

For a root problem

$$
f(x)=0,
$$

the **error** compares the approximation with the exact root:

$$
|x-x^\ast|.
$$

But the exact root may be unknown.

The **residual** measures how well the computed value satisfies the original equation:

$$
|f(x)|.
$$

For

$$
f(x)=x^2-2,
$$

the residual is

$$
|x^2-2|.
$$

````python
x_approx = 1.4142

error = abs(np.sqrt(2) - x_approx)
residual = abs(f(x_approx))

print(f"Error    = {error:.8e}")
print(f"Residual = {residual:.8e}")
````

# 18. Computers Use Finite Precision

A computer cannot represent every real number exactly.

For example, decimal numbers such as

$$
0.1
$$

are stored in binary floating-point form.

This can produce small round-off effects.

````python
print("0.1 + 0.2 =", 0.1 + 0.2)
print("Is 0.1 + 0.2 exactly equal to 0.3?", (0.1 + 0.2) == 0.3)
````

This is **not a Python error**.  
It is a consequence of finite-precision floating-point arithmetic.

# 19. Vectors and Matrices

Chapter 3 will use systems of equations written as

$$
A\mathbf{x}=\mathbf{b}.
$$

NumPy stores vectors and matrices naturally.

````python
A = np.array([
    [4.0, 1.0],
    [2.0, 3.0]
])

b = np.array([1.0, 2.0])

print("A =")
print(A)

print("\nb =")
print(b)
````

A vector

$$
\mathbf{x}
=
\begin{bmatrix}
0.1\\
0.6
\end{bmatrix}
$$

is stored as:

````python
x = np.array([0.1, 0.6])

print("x =", x)
print("A @ x =", A @ x)
````

# 20. Residual of a Linear System

For

$$
A\mathbf{x}=\mathbf{b},
$$

the residual vector is

$$
\mathbf{r}
=
\mathbf{b}-A\mathbf{x}.
$$

If

$$
\mathbf{r}=\mathbf{0},
$$

then the vector exactly satisfies the system.

````python
r = b - A @ x

print("Residual =", r)
print("Residual norm =", np.linalg.norm(r))
````

Here the residual is essentially zero, so the vector is an exact solution up to floating-point round-off.

# 22. Visualizing Convergence

A convergence plot shows how the error changes with iteration.

````python
x = 1.0
errors = []

for k in range(6):
    errors.append(abs(np.sqrt(2) - x))
    x = 0.5 * (x + 2/x)

plt.figure(figsize=(8, 5))
plt.semilogy(range(len(errors)), errors, marker="o")
plt.xlabel("Iteration k")
plt.ylabel("Absolute error (log scale)")
plt.title("Error Decreases During Iteration")
plt.grid(True)
plt.show()
````

A downward trend indicates that the approximations are improving.

Later, convergence plots will help us compare numerical methods.

# 23. Fully Worked Mini-Case

We will solve

$$
x^2-3=0
$$

approximately.

The exact positive root is

$$
\sqrt{3}.
$$

Use the same iterative idea:

$$
x_{k+1}
=
\frac{1}{2}
\left(
x_k+\frac{3}{x_k}
\right).
$$

````python
target = 3.0
x = 2.0
tol = 1e-10
max_iter = 50

print(f"{'k':>3} {'x_k':>15} {'residual':>15}")
print("-" * 36)

for k in range(max_iter):
    residual = abs(x**2 - target)
    print(f"{k:3d} {x:15.10f} {residual:15.6e}")

    x_new = 0.5 * (x + target/x)

    if abs(x_new - x) < tol:
        x = x_new
        break

    x = x_new

print("\nApproximation =", x)
print("np.sqrt(3)    =", np.sqrt(3))
print("Absolute error =", abs(np.sqrt(3) - x))
````

### What this example teaches

A numerical method should include:

1. an initial approximation;
2. an update formula;
3. a loop;
4. a stopping criterion;
5. a maximum iteration limit;
6. a verification step.

This is a reusable template for many algorithms in this course.
